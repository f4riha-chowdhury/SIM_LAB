```python
import numpy as np
from scipy import stats

# Data
data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])

# Significance level
alpha = 0.05

# Sample sizes
n1 = len(data1)
n2 = len(data2)

# Sample means
mean1 = np.mean(data1)
mean2 = np.mean(data2)

# Sample variances (ddof=1)
var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)

# Mean difference
mean_diff = mean1 - mean2

# Pooled variance
pooled_var = ((n1 - 1) * var1 + (n2 - 1) * var2) / (n1 + n2 - 2)

# Pooled standard error
SE = np.sqrt(pooled_var * (1/n1 + 1/n2))

# t-statistic
t_stat = mean_diff / SE

# Degrees of freedom
df = n1 + n2 - 2

# Two-tailed p-value
p_value = 2 * stats.t.sf(abs(t_stat), df)

# Critical t-value
t_critical = stats.t.ppf(1 - alpha/2, df)

# 95% Confidence Interval
CI_lower = mean_diff - t_critical * SE
CI_upper = mean_diff + t_critical * SE

# Display results
print("Two-Sample t-Test (Pooled Equal Variances)")
print("-" * 50)

print(f"Sample mean of Data 1 = {mean1:.3f}")
print(f"Sample mean of Data 2 = {mean2:.3f}")

print(f"Sample variance of Data 1 = {var1:.3f}")
print(f"Sample variance of Data 2 = {var2:.3f}")

print(f"Mean difference (Data 1 - Data 2) = {mean_diff:.3f}")

print(f"Pooled variance = {pooled_var:.4f}")
print(f"Standard Error = {SE:.4f}")

print(f"t-statistic = {t_stat:.3f}")
print(f"Degrees of freedom = {df}")
print(f"Two-tailed p-value = {p_value:.8f}")

print(f"95% Confidence Interval = ({CI_lower:.3f}, {CI_upper:.3f})")

# Decision
if p_value < alpha:
    print("\nDecision: Reject H0")
    print("Conclusion: The population means are significantly different.")
else:
    print("\nDecision: Fail to reject H0")
    print("Conclusion: There is not enough evidence that the population means differ.")
```
