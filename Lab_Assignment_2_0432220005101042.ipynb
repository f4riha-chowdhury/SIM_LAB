**d1 (last digit):** 2  
**d2 (second-last digit):** 4

This notebook investigates the properties of a student-specific matrix and examines how changing exactly one matrix value by +1 affects the determinant, rank, eigenvalues, and inverse.

## 1. Create Personal Matrix

The assignment defines:

- `d1` = last digit of the student ID
- `d2` = second-last digit of the student ID

For Student ID **0432220005101042**:

- `d1 = 2`
- `d2 = 4`

Using the required matrix:

\[
A =
\begin{bmatrix}
d1+2 & d2+1 \\
2d1 & d2+2
\end{bmatrix}
=
\begin{bmatrix}
4 & 5 \\
4 & 6
\end{bmatrix}
\]

```python
import numpy as np

# Student ID
student_id = "0432220005101042"

# Extract d1 and d2
d1 = int(student_id[-1])
d2 = int(student_id[-2])

# Create personal matrix A
A = np.array([
    [d1 + 2, d2 + 1],
    [2 * d1, d2 + 2]
], dtype=float)

print("Student ID:", student_id)
print("d1 =", d1)
print("d2 =", d2)
print("\nMatrix A:")
print(A)
```

## 2. Analyze Matrix A

```python
# Calculate properties of Matrix A
shape_A = A.shape
det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
eigenvalues_A = np.linalg.eigvals(A)

print("Shape of A:", shape_A)
print("Determinant of A:", det_A)
print("Rank of A:", rank_A)
print("Eigenvalues of A:", eigenvalues_A)

if not np.isclose(det_A, 0):
    inverse_A = np.linalg.inv(A)
    print("\nInverse of A:")
    print(inverse_A)
else:
    print("\nA is not invertible because its determinant is zero.")
```

### Results for Matrix A

- **Shape:** `(2, 2)`
- **Determinant:** `4`
- **Rank:** `2`
- **Eigenvalues:** approximately `0.4174` and `9.5826`
- **Inverse:** exists because the determinant is non-zero.

## 3. Value Change Experiment

According to the assignment, exactly **one value** of matrix A must be increased by **+1**.

Here, the element at **row 1, column 1** is changed:

\[
A_{1,1}: 4 \rightarrow 5
\]

Therefore:

\[
B =
\begin{bmatrix}
5 & 5 \\
4 & 6
\end{bmatrix}
\]

```python
# Create Matrix B by adding +1 to A[0, 0]
B = A.copy()
B[0, 0] += 1

print("Changed element: A[0, 0] (row 1, column 1)")
print("Old value:", A[0, 0])
print("New value:", B[0, 0])

print("\nMatrix B:")
print(B)
```

## 4. Re-analyze Matrix B

```python
# Calculate properties of Matrix B
shape_B = B.shape
det_B = np.linalg.det(B)
rank_B = np.linalg.matrix_rank(B)
eigenvalues_B = np.linalg.eigvals(B)

print("Shape of B:", shape_B)
print("Determinant of B:", det_B)
print("Rank of B:", rank_B)
print("Eigenvalues of B:", eigenvalues_B)

if not np.isclose(det_B, 0):
    inverse_B = np.linalg.inv(B)
    print("\nInverse of B:")
    print(inverse_B)
else:
    print("\nB is not invertible because its determinant is zero.")
```

### Results for Matrix B

- **Shape:** `(2, 2)`
- **Determinant:** `10`
- **Rank:** `2`
- **Eigenvalues:** `1` and `10`
- **Inverse:** exists because the determinant is non-zero.

## 5. Comparison and Explanation

### 1. How did the determinant change and why?

The determinant increased from **4 to 10**. This happened because one element of matrix A was increased by 1, which changed the determinant calculation.

### 2. Did the rank change? Explain.

No. The rank remained **2**. Both matrices have two linearly independent rows, so changing the selected value did not reduce the rank.

### 3. How did the eigenvalues respond to the value change?

The eigenvalues changed from approximately **0.4174 and 9.5826** for A to **1 and 10** for B. This shows that changing even one matrix value can change its eigenvalues.

### 4. Is B easier or harder to invert than A? Why?

Both A and B are invertible because their determinants are non-zero. Matrix B has a determinant of **10**, compared with **4** for A, so B is farther from having a zero determinant and is generally less close to being singular.
