```python
import numpy as np
import matplotlib.pyplot as plt


# Scalar
scalar = 3.7

print("Scalar Value =", scalar)

print("sin =", np.sin(scalar))
print("cos =", np.cos(scalar))
print("tan =", np.tan(scalar))

print("arcsin =", np.arcsin(np.clip(np.sin(scalar), -1, 1)))
print("arccos =", np.arccos(np.clip(np.cos(scalar), -1, 1)))
print("arctan =", np.arctan(scalar))

print("exp =", np.exp(scalar))
print("log =", np.log(scalar))
print("log10 =", np.log10(scalar))
print("sqrt =", np.sqrt(scalar))
print("square =", np.square(scalar))
print("absolute =", np.abs(-scalar))

print("round =", np.round(scalar))
print("floor =", np.floor(scalar))
print("ceil =", np.ceil(scalar))

# array
arr = np.array([2.5, 4.8, 6.1, 8.3, 10.7, 12.4])

print("Array =", arr)

print("sin =", np.sin(arr))
print("cos =", np.cos(arr))
print("tan =", np.tan(arr))

print("exp =", np.exp(arr))
print("log =", np.log(arr))
print("log10 =", np.log10(arr))

print("sqrt =", np.sqrt(arr))
print("square =", np.square(arr))
print("absolute =", np.abs(-arr))

print("round =", np.round(arr))
print("floor =", np.floor(arr))
print("ceil =", np.ceil(arr))

# matrix
matrix = np.array([
    [1.5, 2.8],
    [3.2, 4.6],
    [5.9, 6.4]
])

print("Original Matrix")
print(matrix)

#funcitons

print("Sin")
print(np.sin(matrix))

print("Cos")
print(np.cos(matrix))

print("Tan")
print(np.tan(matrix))

print("Arcsin")
print(np.arcsin(np.clip(matrix/7, -1, 1)))

print("Arccos")
print(np.arccos(np.clip(matrix/7, -1, 1)))

print("Arctan")
print(np.arctan(matrix))

print("Exponential")
print(np.exp(matrix))

print("Natural Log")
print(np.log(matrix))

print("Log10")
print(np.log10(matrix))

# vector funciton matrix 2*3

A = np.array([
    [2.1,4.7,1.8],
    [6.4,5.5,3.2]
])

print("Matrix")
print(A)

print("Maximum =", np.max(A))
print("Maximum Index =", np.argmax(A))

print("Minimum =", np.min(A))
print("Minimum Index =", np.argmin(A))

print("Length =", len(A.flatten()))

print("Row-wise Descending")
print(np.sort(A,axis=1)[:,::-1])

print("Entire Matrix Descending")
print(np.sort(A.flatten())[::-1])

print("Sum =", np.sum(A))
print("Product =", np.prod(A))

print("Median =", np.median(A))
print("Mean =", np.mean(A))
print("Standard Deviation =", np.std(A))

random_matrix = np.random.uniform(1,10,(2,3))

print(random_matrix)
```
